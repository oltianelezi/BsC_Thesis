```python
# Import libraries, check dependencies

import torch
from ultralytics import YOLO

print(torch.__version__)
print(torch.cuda.is_available())  # should print True
print(torch.cuda.get_device_name(0))  # should show RTX 3060
```

```python
# Training YoloV8s

v8s_model = YOLO("yolov8s.pt")  # downloads pretrained weights automatically

v8s_model.train(
    data="football-players-detection-1/data.yaml",
    device="cuda",
    imgsz=1280,
    batch=6,
    epochs=25,
    optimizer="AdamW",
    project="football_detection",
    name="yolov8s_1280",
)
```
