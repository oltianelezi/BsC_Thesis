```python
# Import libraries, check dependencies

import torch
from ultralytics import YOLO

print(torch.__version__)
print("MPS available:", torch.backends.cuda.is_available())
```

```python
# Training YoloV8s

v8s_model = YOLO("yolov8s.pt")  # downloads pretrained weights automatically

v8s_model.train(
    data="football-players-detection-1/data.yaml",
    device="cuda",
    imgsz=1280,
    batch=6,
    epochs=25,
    optimizer="AdamW",
    project="football_detection",
    name="yolov8s_1280",
)
```
